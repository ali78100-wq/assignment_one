```python
text_cols=df.select_dtypes(include='str').columns

df[text_cols]=df[text_cols].apply(lambda x: x.str.strip())
```

```python
for col in [
    "Ship Mode", "Segment", "Category", "Sub-Category",
    "Region", "State", "City"
]: df[col]=df[col].str.title()
```

```python
for col in ["Order Date", "Ship Date"]:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")


df[["Order Date", "Ship Date"]].head()
```

```python
for col in ["Sales", "Profit", "Discount", "Quantity","Postal Code"]:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

df[["Sales", "Profit", "Discount", "Quantity","Postal Code"]].head().dtypes
```

```python
def find_outliers_iqr(data, column):
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    mask = (data[column] < lower) | (data[column] > upper)
    return lower, upper, data.loc[mask, column]

for col in ["Sales", "Profit", "Discount", "Quantity"]:
    if col in df.columns:
        lower, upper, outliers = find_outliers_iqr(df, col)
        print(f"\\n{col}:")
        print("الحد الأدنى:", round(lower, 3))
        print("الحد الأعلى:", round(upper, 3))
        print("عدد القيم الشاذة:", len(outliers))
        display(outliers.head(10))

# افحص القيم الشاذة قبل اتخاذ قرار بحذفها أو استبدالها.
```
